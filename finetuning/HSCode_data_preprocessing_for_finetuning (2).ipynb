# Final dataset

```python
import json

records = []

for _, row in df.iterrows():
    old_hs = str(row.get("OldHSCode", "")).strip()
    new_hs = str(row.get("NewHSCode", "")).strip()
    desc_en = str(row.get("LongDescEn", "")).strip()

    # Numeric fields: convert safely
    def to_float(val):
        try:
            return float(val)
        except:
            return 0.0

    unit = str(row.get("StatisticalQtyUnit", "")).strip()
    duty = to_float(row.get("DutyPercentage", 0))
    alt_duty = to_float(row.get("AltDutyPercentage", 0))
    service = to_float(row.get("ServiceCharge", 0))
    exemption = to_float(row.get("Exemption", 0))
    spec_duty = to_float(row.get("SpecificDuty", 0))
    spec_qty = to_float(row.get("SpecificDutyQty", 0))

    # Assistant output as CLEAN JSON
    assistant_json = {
        "NewHSCode": new_hs,
        "OldHSCode": old_hs,
        "StatisticalQtyUnit": unit,
        "DutyPercentage": duty,
        "AltDutyPercentage": alt_duty,
        "ServiceCharge": service,
        "Exemption": exemption,
        "SpecificDuty": spec_duty,
        "SpecificDutyQty": spec_qty
    }

    sample = {
        "messages": [
            {
                "role": "system",
                "content": (
                    "You are an HS Code classification assistant. "
                    "Always provide the New HS Code clearly and accurately along with related attributes."
                )
            },
            {
                "role": "user",
                "content": f"Item description: {desc_en}. Provide all HS details in JSON format."
            },
            {
                "role": "assistant",
                "content": json.dumps(assistant_json, ensure_ascii=False)
            }
        ]
    }

    records.append(sample)

output_file = "train.jsonl"

with open(output_file, "w", encoding="utf-8") as f:
    for r in records:
        f.write(json.dumps(r, ensure_ascii=False) + "\n")

print(f"✅ Fine-tuning dataset created successfully: {output_file}")
print(f"Total samples: {len(records)}")
```

Encoded hs code

```python
import json
import pandas as pd

# 2. BUILD ENCODED ID MAP (5-digit sequential numbering)

hs_to_id = {}
id_to_hs = {}

current_id = 1

for _, row in df.iterrows():
    hs_code = str(row.get("NewHSCode", "")).strip()

    # skip invalid rows
    if len(hs_code) == 0:
        continue

    # Assign ID only if not already assigned
    if hs_code not in hs_to_id:
        encoded = str(current_id).zfill(5)    # "00001"
        hs_to_id[hs_code] = encoded
        id_to_hs[encoded] = hs_code
        current_id += 1


# Save lookup tables
with open("hs_to_id.json", "w", encoding="utf-8") as f:
    json.dump(hs_to_id, f, ensure_ascii=False, indent=2)

with open("id_to_hs.json", "w", encoding="utf-8") as f:
    json.dump(id_to_hs, f, ensure_ascii=False, indent=2)

print("✅ Lookup tables created: hs_to_id.json, id_to_hs.json")
print(f"Total unique HS codes: {len(hs_to_id)}")


# 3. CREATE TRAINING DATA WITH ENCODED HSID
records = []

for _, row in df.iterrows():
    desc_en = str(row.get("LongDescEn", "")).strip()
    hs_code = str(row.get("NewHSCode", "")).strip()

    if hs_code not in hs_to_id:
        continue

    encoded_id = hs_to_id[hs_code]   # "00001"

    sample = {
        "messages": [

            {
                "role": "user",
                "content": f"Item description: {desc_en}"
            },
            {
                "role": "assistant",
                "content": f"<HSID>{encoded_id}</HSID>"
            }
        ]
    }

    # Duplicate the sample for more training weight (optional)
    records.append(sample)
    # records.append(sample.copy())   # enable if you want duplication


# Save final JSONL
output_file = "train.jsonl"

with open(output_file, "w", encoding="utf-8") as f:
    for r in records:
        f.write(json.dumps(r, ensure_ascii=False) + "\n")

print(f"✅ Encoded fine-tuning dataset saved: {output_file}")
print(f"Total samples: {len(records)}")
```

# Bilingual dataset

```python
import pandas as pd
import json

# Create bilingual training pairs
records = []

for _, row in df.iterrows():
    new_hscode = str(row["NewHSCode"]).strip()
    desc_en = str(row["LongDescEn"]).strip()
    desc_ar = str(row["LongDescAr"]).strip()
    unit = str(row["StatisticalQtyUnit"]).strip()
    duty = str(row["DutyPercentage"]).strip()
    alt_duty = str(row["AltDutyPercentage"]).strip()

    # ---- English version ----
    user_en = {
        "role": "user",
        "content": f"What is the HS Code for {desc_en}?"
    }
    assistant_en = {
        "role": "assistant",
        "content": (
            f"HS Code -> {new_hscode}\n"
            f"Reasoning -> The item '{desc_en}' is classified under HS Code {new_hscode} "
            f"based on its description and category. "
            f"Duty: {duty}%, Alt Duty: {alt_duty}%, Unit: {unit}."
        )
    }
    records.append([user_en, assistant_en])

    # ---- Arabic version ----
    user_ar = {
        "role": "user",
        "content": f"ما هو رمز النظام المنسق (HS Code) لـ{desc_ar}؟"
    }
    assistant_ar = {
        "role": "assistant",
        "content": (
            f"رمز النظام المنسق -> {new_hscode}\n"
            f"السبب -> يُصنف البند '{desc_ar}' تحت رمز النظام المنسق {new_hscode} "
            f"وفقًا لوصفه وفئته. الرسوم الجمركية: {duty}٪، الرسوم البديلة: {alt_duty}٪، وحدة القياس: {unit}."
        )
    }
    records.append([user_ar, assistant_ar])

# Save as JSON
output_file = "bilingual_finetune_dataset.json"
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(records, f, ensure_ascii=False, indent=2)

print(f"✅ Bilingual fine-tuning dataset created successfully: {output_file}")
print(f"Total samples: {len(records)}")
```
